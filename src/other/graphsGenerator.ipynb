```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

configPath = "../config.cfg"

def plot_mean_max_by_size(df, x_col, y_col, size_col):
    grouped = df.groupby([x_col, size_col])[y_col].agg(['mean', 'max']).reset_index()

    print(f"{y_col} vs {x_col}: ")
    plt.figure()

    sizes = sorted(grouped[size_col].unique())
    colors = plt.cm.tab10(np.linspace(0, 1, len(sizes)))

    for size, color in zip(sizes, colors):
        subset = grouped[grouped[size_col] == size]

        plt.plot(
            subset[x_col], subset['mean'],
            marker='o',
            linestyle='-',
            color=color,
            label=f"size={size} (mean)"
        )
        plt.plot(
            subset[x_col], subset['max'],
            marker='x',
            linestyle='--',
            color=color,
            alpha=0.7,
            label=f"size={size} (max)"
        )

    plt.xlabel(x_col)
    plt.ylabel(y_col)
    plt.title(f"{y_col} vs {x_col} by {size_col}")
    plt.legend()
    plt.grid(True)
    plt.show()

def plot_frequency(df, x_col):
    counts = df[x_col].value_counts().sort_index()
    plt.figure()
    plt.bar(counts.index, counts.values)
    plt.xlabel(x_col)
    plt.ylabel("Frequency")
    plt.title(f"Frequency of {x_col}")
    plt.grid(True)
    plt.show()

def plot_mean_max(df, x_col, y_col):
    grouped = df.groupby(x_col)[y_col]
    
    mean_vals = grouped.mean()
    max_vals = grouped.max()

    # check if both are constant
    if mean_vals.nunique() == 1 and max_vals.nunique() == 1:
        print(f"{y_col} vs {x_col}: constant values (mean={mean_vals.iloc[0]}, max={max_vals.iloc[0]})")
    else:
        print(f"{y_col} vs {x_col}: ")
        plt.figure()
        
        plt.plot(mean_vals.index, mean_vals.values, marker='o', label='Mean')
        plt.plot(max_vals.index, max_vals.values, marker='x', label='Max')
        
        plt.xlabel(x_col)
        plt.ylabel(y_col)
        plt.title(f"{y_col} vs {x_col} (mean & max)")
        plt.legend()
        plt.grid(True)
        plt.show()

def plot_avg(df, x_col, y_col):
    grouped = df.groupby(x_col)[y_col].mean()

    # constant value check to avoid making not interesting graphs
    if grouped.nunique() == 1:
        value = grouped.iloc[0]
        print(f"{y_col} vs {x_col}: constant value {value}")
    else:
        print(f"{y_col} vs {x_col}: ")
        plt.figure()
        plt.plot(grouped.index, grouped.values, marker='o')
        plt.xlabel(x_col)
        plt.ylabel(f"Average {y_col}")
        plt.title(f"{y_col} vs {x_col}")
        plt.grid(True)
        plt.show()

def plot_mean_by_size(df, x_col, y_col, size_col):
    grouped = df.groupby([x_col, size_col])[y_col].agg(['mean']).reset_index()

    if grouped['mean'].nunique() == 1:
        value = grouped['mean'].iloc[0]
        print(f"{y_col} vs {x_col}: constant value {value} across all {size_col}")
        return

    print(f"{y_col} vs {x_col}:")
    plt.figure()

    sizes = sorted(grouped[size_col].unique())
    colors = plt.cm.tab10(np.linspace(0, 1, len(sizes)))

    for size, color in zip(sizes, colors):
        subset = grouped[grouped[size_col] == size]

        plt.plot(
            subset[x_col],
            subset['mean'],
            marker='o',
            linestyle='-',
            color=color,
            label=f"{size_col}={size}"
        )

    plt.xlabel(x_col)
    plt.ylabel(y_col)
    plt.title(f"{y_col} vs {x_col} by {size_col}")
    plt.legend(title=size_col)
    plt.grid(True)
    plt.show()


possibleX = ["maxTreeSize", "maxCycleSize", "maxDistanceToCycle", "sinkCount", "sourceCount", "isolatedCount"]
distances = ["hammingDist_AV_to_SP","hammingDist_AV_to_CC","hammingDist_AV_to_PAV","hammingDist_AV_to_MES_AV","hammingDist_AV_to_GJCR_AV",
             "hammingDist_SP_to_CC","hammingDist_SP_to_PAV","hammingDist_SP_to_MES_AV","hammingDist_SP_to_GJCR_AV",
             "hammingDist_CC_to_PAV","hammingDist_CC_to_MES_AV","hammingDist_CC_to_GJCR_AV",
             "hammingDist_PAV_to_MES_AV","hammingDist_PAV_to_GJCR_AV",
             "hammingDist_MES_AV_to_GJCR_AV"]

def load_config(path):
    cfg = {}

    with open(path) as f:
        for line in f:
            line = line.strip()

            if not line or line.startswith("#"):
                continue

            if "=" not in line:
                continue

            key, value = line.split("=", 1)
            key = key.strip()
            value = value.strip().lower()

            # bool hodnoty
            if value in ("true", "1"):
                cfg[key] = True
            elif value in ("false", "0"):
                cfg[key] = False
            else:
                cfg[key] = value  # string / číslo

    return cfg

cfg = load_config(configPath)
possibleY = []

if cfg["JR"]: possibleY += ["JR_AV","JR_SP","JR_CC","JR_PAV","JR_MES","JR_GJCR"]
if cfg["PJR"]: possibleY += ["PJR_AV","PJR_SP","PJR_CC","PJR_PAV","PJR_MES","PJR_GJCR"]
if cfg["EJR"]: possibleY += ["EJR_AV","EJR_SP","EJR_CC","EJR_PAV","EJR_MES","EJR_GJCR"]
if cfg["EJRplus"]: possibleY += ["EJR+_AV","EJR+_SP","EJR+_CC","EJR+_PAV","EJR+_MES","EJR+_GJCR"]
if cfg["IR"]: possibleY += ["IR_AV","IR_SP","IR_CC","IR_PAV","IR_MES","IR_GJCR"]
if cfg["LR"]: possibleY += ["LR_AV","LR_SP","LR_CC","LR_PAV","LR_MES","LR_GJCR"]
if cfg["PR"]: possibleY += ["PR_AV","PR_SP","PR_CC","PR_PAV","PR_MES","PR_GJCR"]
if cfg["CS"]: possibleY += ["CS_AV","CS_SP","CS_CC","CS_PAV","CS_MES","CS_GJCR"]
if cfg["sJR"]: possibleY += ["sJR_AV","sJR_SP","sJR_CC","sJR_PAV","sJR_MES","sJR_GJCR"]
if cfg["sEJR"]: possibleY += ["sEJR_AV","sEJR_SP","sEJR_CC","sEJR_PAV","sEJR_MES","sEJR_GJCR"]
if cfg["sEJRplus"]: possibleY += ["sEJR+_AV","sEJR+_SP","sEJR+_CC","sEJR+_PAV","sEJR+_MES","sEJR+_GJCR"]

df = pd.read_csv("../results.csv")

for x in possibleX:
    plot_frequency(df, x)

for d in distances:
    for x in possibleX:
        plot_mean_max_by_size(df, x, d, "committeeSize")

#for d in distances:
#    for x in possibleX:
#        plot_mean_max(df, x, d)

for y in possibleY:
    for x in possibleX:
        plot_mean_by_size(df, x, y, "committeeSize")
```
