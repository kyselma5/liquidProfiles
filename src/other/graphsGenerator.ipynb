```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_frequency(df, x_col):
    counts = df[x_col].value_counts().sort_index()
    plt.figure()
    plt.bar(counts.index, counts.values)
    plt.xlabel(x_col)
    plt.ylabel("Frequency")
    plt.title(f"Frequency of {x_col}")
    plt.grid(True)
    plt.show()

def plot_avg(df, x_col, y_col):
    grouped = df.groupby(x_col)[y_col].mean()

    # constant value check to avoid making not interesting graphs
    if grouped.nunique() == 1:
        value = grouped.iloc[0]
        print(f"{y_col} vs {x_col}: constant value {value}")
    else:
        print(f"{y_col} vs {x_col}: ")
        plt.figure()
        plt.plot(grouped.index, grouped.values, marker='o')
        plt.xlabel(x_col)
        plt.ylabel(f"Average {y_col}")
        plt.title(f"{y_col} vs {x_col}")
        plt.grid(True)
        plt.show()

possibleX = ["maxTreeSize", "maxCycleSize", "maxDistanceToCycle", "sinkCount", "sourceCount", "isolatedCount"]
possibleY = [       
            "hammingDist_AV_to_SP_AV",
            "hammingDist_AV_to_CC_AV",
            "hammingDist_AV_to_PAV_AV",
            "hammingDist_AV_to_MES_AV",
            "hammingDist_AV_to_GJCR_AV",
             "JR_AV","JR_SP","JR_CC","JR_PAV","JR_MES","JR_GJCR",
             "EJR_AV","EJR_SP","EJR_CC","EJR_PAV","EJR_MES","EJR_GJCR",
             "EJR+_AV","EJR+_SP","EJR+_CC","EJR+_PAV","EJR+_MES","EJR+_GJCR",
             "PR_AV","PR_SP","PR_CC","PR_PAV","PR_MES","PR_GJCR",
             "CS_AV","CS_SP","CS_CC","CS_PAV","CS_MES","CS_GJCR",
             "SJR_AV","SJR_SP","SJR_CC","SJR_PAV","SJR_MES","SJR_GJCR",
             "SEJR+_AV","SEJR+_SP","SEJR+_CC","SEJR+_PAV","SEJR+_MES","SEJR+_GJCR"]

df = pd.read_csv("../results.csv")

for x in possibleX:
    plot_frequency(df, x)
    
for x in possibleX:
    for y in possibleY:
        plot_avg(df, x, y)
```
